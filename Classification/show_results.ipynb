```python
%matplotlib widget
import pandas as pd
from IPython.display import display

BASE_DIR = "./Results_StdSubject_AllTasks/20251016_MinOvSamp"

BINARY = f"{BASE_DIR}/20251016-194311_binary_10s&0.2_RFEf.21.csv"
TERNARY = f"{BASE_DIR}/20251016-195254_ternary_10s&0.2_RFEf.21.csv"
QUATERNARY = f"{BASE_DIR}/20251016-201512_quaternary_10s&0.2_RFEf.21.csv"
```

```python
pd_bin = pd.read_csv(BINARY, sep=";", index_col=0)
display(pd_bin.groupby(["classifier"]).mean()[["f1-score", "accuracy", "time"]])
display(pd_bin.groupby(["classifier"]).std()[["f1-score", "accuracy", "time"]])

pd_ter = pd.read_csv(TERNARY, sep=";", index_col=0)
display(pd_ter.groupby(["classifier"]).mean()[["f1-score", "accuracy", "time"]])
display(pd_ter.groupby(["classifier"]).std()[["f1-score", "accuracy", "time"]])

pd_qad = pd.read_csv(QUATERNARY, sep=";", index_col=0)
display(pd_qad.groupby(["classifier"]).mean()[["f1-score", "accuracy", "time"]])
display(pd_qad.groupby(["classifier"]).std()[["f1-score", "accuracy", "time"]])
```
